```python
import pandas as pd

# Define the dictionary with data for four students
data = {
    'Name': ['Aarav', 'Diya', 'Kabir', 'Ananya'],
    'Age': [20, 21, 19, 22],
    'City': ['Delhi', 'Mumbai', 'Bangalore', 'Chennai']
}

# Create the DataFrame
student_df = pd.DataFrame(data)

# Display the DataFrame
print(student_df)
```

```python
import pandas as pd

# Create a dictionary of equal-length lists
book_data = {
    'Book': ['Python 101', 'Data Science Guide', 'Machine Learning Basics'],
    'Author': ['Alice Smith', 'Bob Jones', 'Charlie Brown'],
    'Price': [29.99, 45.50, 39.99]
}

# Create the book_df DataFrame
book_df = pd.DataFrame(book_data)
print(book_df)
```

```python
# Assign the same publisher name to every row in book_df
book_df['Publisher'] = 'TechPress'

print(book_df)
```

```python
import numpy as np

# Create a five-row DataFrame with some base data
df = pd.DataFrame({
    'Item': ['Item A', 'Item B', 'Item C', 'Item D', 'Item E']
})

# Assign values from np.arange(5) to a new column named SerialNo
df['SerialNo'] = np.arange(5)

print(df)
```

```python
import pandas as pd

# Create the movie_df DataFrame
movie_df = pd.DataFrame({
    'Movie': ['Inception', 'Interstellar', 'The Dark Knight'],
    'Year': [2010, 2014, 2008],
    'Rating': [8.8, 8.7, 9.0]
})

print(movie_df)
```

```python
# Using assign to add the new 'Director' column with NaN or placeholder values
new_movie_df = movie_df[['Movie', 'Rating']].assign(Director=np.nan)

print(new_movie_df)
```

```python
# Create a Series with city names at index 0 and 3
city_series = pd.Series(['Delhi', 'Mumbai'], index=[0, 3])

print(city_series)
```

```python
# Create a sample 4-row DataFrame (indexes 0, 1, 2, 3)
four_row_df = pd.DataFrame({
    'Student': ['Aarav', 'Diya', 'Kabir', 'Ananya']
})

# Assign the Series (it will align correctly at index 0 and 3, placing NaN at 1 and 2)
four_row_df['Branch'] = city_series

print(four_row_df)
```

```python
# Read tabular data directly from your system clipboard into a DataFrame
copied_df = pd.read_clipboard()

print(copied_df)
```

```python
import pandas as pd

# Create a 5-row employee DataFrame
employee_df = pd.DataFrame({
    'Employee': ['Aarav', 'Diya', 'Kabir', 'Ananya', 'Rohan'],
    'Department': ['HR', 'IT', 'Finance', 'Marketing', 'Sales'],
    'Experience': [3, 5, 2, 4, 6]
})

# Create a new DataFrame with only Employee and Experience columns
subset_df = employee_df[['Employee', 'Experience']]
print(subset_df)
```

```python
# Retrieve the Department column using one-column access
department_col = employee_df['Department']
print(department_col)

# Display all column names of employee_df
print(employee_df.columns)
```
